```python
import fastf1 as ff1
from fastf1 import plotting
from matplotlib import pyplot as plt

%matplotlib widget

ff1.plotting.setup_mpl(mpl_timedelta_support=True, color_scheme="fastf1")
```

```python
race = ff1.get_session(2023, "Azerbaijan", "R")
race.load()
```

```python
fig, ax = plt.subplots(figsize=(8, 5))

for driver in ('HAM', 'PER', 'VER', 'RUS'):
    laps = race.laps.pick_drivers(driver).reset_index()
    style = plotting.get_driver_style(identifier=driver,
                                      style=['color', 'linestyle'],
                                      session=race)
    ax.plot(laps['LapTime'], **style, label=driver)

ax.set_xlabel("Lap Number")
ax.set_ylabel("Lap Time")
ax.legend()
plotting.add_sorted_driver_legend(ax, race)
```
